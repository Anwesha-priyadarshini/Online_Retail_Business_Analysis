```python
customer_dates = (
    df_product_sales.groupby("CustomerID")["InvoiceDate"]
    .agg(["min", "max"])
)

customer_dates["purchase_span_days"] = (
    customer_dates["max"] - customer_dates["min"]
).dt.days

customer_dates.head()
```

```python
import sqlite3

conn = sqlite3.connect("online_retail.db")

df_product_sales.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)

print("SQL database created successfully!")
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(DISTINCT CustomerID) AS total_customers,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_units,
    ROUND(SUM(Revenue), 2) AS total_revenue
FROM sales;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Country,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_units,
    ROUND(SUM(Revenue), 2) AS total_revenue
FROM sales
GROUP BY Country
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    Description,
    SUM(Quantity) AS total_units,
    ROUND(SUM(Revenue), 2) AS total_revenue
FROM sales
GROUP BY Description
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    CustomerID,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    SUM(Quantity) AS total_units,
    ROUND(SUM(Revenue), 2) AS total_revenue
FROM sales
WHERE CustomerID IS NOT NULL
GROUP BY CustomerID
ORDER BY total_revenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    SUBSTR(InvoiceDate, 1, 7) AS month,
    ROUND(SUM(Revenue), 2) AS total_revenue
FROM sales
GROUP BY month
ORDER BY month;
"""

monthly_sql = pd.read_sql_query(query, conn)

monthly_sql
```

```python
query = """
SELECT
    CASE
        WHEN order_count = 1 THEN 'One-time customer'
        ELSE 'Repeat customer'
    END AS customer_type,
    COUNT(*) AS number_of_customers
FROM (
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS order_count
    FROM sales
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
)
GROUP BY customer_type
ORDER BY number_of_customers DESC;
"""

customer_type_sql = pd.read_sql_query(query, conn)

customer_type_sql
```

```python
query = """
SELECT
    Country,
    COUNT(DISTINCT InvoiceNo) AS total_orders,
    ROUND(SUM(Revenue), 2) AS total_revenue,
    ROUND(SUM(Revenue) / COUNT(DISTINCT InvoiceNo), 2) AS average_order_value
FROM sales
GROUP BY Country
HAVING COUNT(DISTINCT InvoiceNo) >= 20
ORDER BY average_order_value DESC;
"""

country_aov_sql = pd.read_sql_query(query, conn)

country_aov_sql
```
